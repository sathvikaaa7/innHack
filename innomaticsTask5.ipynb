```python
#Problem 1: Social Media – Post Engagement Analyzer
def post_eng(l):
    t = 0
    for i in l:
        t += i

    if t >= 1000:
        s = "Viral Post"
    else:
        s = "Normal Engagement"

    print("Total Likes:", t)
    print("Post Status:", s)

post_eng([200, 300, 550])
```

```python
#Problem 2: Healthcare – Medicine Stock Alert System
def med_stock(s):
    if s < 10:
        st = "Low Stock Alert"
    else:
        st = "Stock Sufficient"

    print("Medicine Stock:", s)
    print("Status:", st)

med_stock(6)
```

```python
#Problem 3: Agriculture – Rainfall Adequacy Checker
def rain_chk(r, req):
    tot = 0
    for i in r:
        tot += i

    avg = tot / len(r)

    if avg >= req:
        st = "Adequate Rainfall"
    else:
        st = "Inadequate Rainfall"

    print("Average Rainfall:", avg)
    print("Rainfall Status:", st)

rain_chk([70, 75, 71], 70)
```

```python
#Problem 4: Social Media – Duplicate Account Detection
def dup_acc(u):
    if len(u) != len(set(u)):
        print("Duplicate Accounts Found: Yes")
    else:
        print("Duplicate Accounts Found: No")

dup_acc(["john", "emma", "john"])
```

```python
#Problem 5: Healthcare – Appointment Eligibility Checker
def appt_chk(a):
    if a >= 18:
        st = "Eligible"
    else:
        st = "Not Eligible"

    print("Patient Age:", a)
    print("Eligibility Status:", st)

appt_chk(21)
```

```python
#Problem 6: Agriculture – Premium Crop Price Filter
def prem_crop(p):
    pr = []
    for i in p:
        if i > 2000:
            pr.append(i)

    print("Premium Crops:", pr)

prem_crop([1500, 2500, 1800, 3200])
```

```python
#Problem 7: System Monitoring – Application Health Checker
def sys_chk(e):
    if e == 0:
        st = "Healthy"
    elif e <= 5:
        st = "Minor Issues"
    else:
        st = "Critical Issues"

    print("Error Count:", e)
    print("System Status:", st)

sys_chk(7)
```

```python
#Problem 8: Banking – Daily Transaction Limit Checker
def txn_chk(a):
    if a <= 50000:
        st = "Approved"
    else:
        st = "Rejected"

    print("Transaction Amount:", a)
    print("Transaction Status:", st)

txn_chk(60000)
```

```python
#Problem 9: E-Learning – Student Attendance Eligibility System
def att_chk(att):
    t = len(att)
    p = 0

    for i in att:
        if i == 1:
            p += 1

    per = (p / t) * 100

    if per >= 75:
        st = "Eligible"
    else:
        st = "Not Eligible"

    print("Attendance Percentage:", per)
    print("Exam Eligibility:", st)

att_chk([1, 1, 1, 0, 1])
```

```python
#Problem 10: Smart Electricity Bill Analyzer
def ele_bill(u):
    b = 0

    if u <= 100:
        b = u * 3
    elif u <= 200:
        b = (100 * 3) + ((u - 100) * 5)
    else:
        b = (100 * 3) + (100 * 5) + ((u - 200) * 7)

    if b < 500:
        st = "Low Usage"
    elif b <= 1500:
        st = "Moderate Usage"
    else:
        st = "High Usage"

    print("Total Bill:", b)
    print("Usage Status:", st)

ele_bill(250)
```
